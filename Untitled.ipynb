```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv")
dataset
```

```python
y=dataset['medv']
x=dataset[['age']]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=10)
```

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
result=lr.predict(x_test)
mse=mean_squared_error(y_test,result)
rmse=np.sqrt(mse)
r2=r2_score(y_test,result)
```

```python
print("Intercept:",lr.intercept_)
print("coefficient:",lr.coef_[0])
print(f" Mean Squared Error: {mse:.2f}")
print(f" Root Mean Squared Error:{rmse:.2f}")
print(f" R2Score:{r2:.2f}")
```

```python
plt.figure(figsize=(6,5))
plt.scatter(y_test,result)
plt.plot([min(y_test),max(y_test)],[min(y_test),max(y_test)],color="red",linestyle="--")
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Linear Regression")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
df=pd.read_csv(r"c")
df.head()
```

```python
plt.scatter(df.age,df.bought_insurance,marker='+',color='red')
plt.show()
```

```python
x_train,x_test,y_train,y_test=train_test_split(df[['age']],df.bought_insurance,train_size=0.8)
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
y_predicted=model.predict(x_test)
model.predict_proba(x_test)
model.score(x_test,y_test)
classification_rep=classification_report(y_test,y_predicted)
print(classification_rep)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data=pd.read_csv(r"C:\Users\anilp\Downloads\iris.csv")
data
```

```python
x=data.drop('species',axis=1).values
```

```python
kmeans=KMeans(n_clusters=3,random_state=42)
kmeans.fit(x)
```

```python
cl=kmeans.labels_
ce=kmeans.cluster_centers_
```

```python
plt.figure(figsize=(10,6))
plt.scatter(x[:,0],x[:,1],c=cl,cmap="viridis",s=50)
plt.scatter(ce[:,0],ce[:,1],c="red",marker='x',s=250)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("K-Means Clustering")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score,classification_report
```

```python
data=pd.read_csv(r"C:\Users\anilp\Downloads\iris.csv")
data
```

```python
x=data.drop('species',axis=1)
y=data['species']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
dt=DecisionTreeClassifier(criterion='entropy',splitter='random',max_depth=3)
dt.fit(x_train,y_train)
```

```python
y_pred=dt.predict(x_test)
acc=accuracy_score(y_test,y_pred)
print("Accuracy:",acc)
print("\n Classification Report")
print(classification_report(y_test,y_pred))
```

```python
class_names_list=y.unique().astype(str).tolist()
```

```python
feature_names_list=x.columns.tolist()
plt.figure(figsize=(10,12))
plot_tree(dt,filled=True,feature_names=feature_names_list,class_names=class_names_list,rounded=True)
plt.title("Decision Tree Visualizations")
plt.show()
```
